```
%%sql -r dataframe_2
USE DATABASE anycompany_lab;
USE SCHEMA SILVER;
```

## 1. CUSTOMER_DEMOGRAPHICS_CLEAN
### Périmètre métier
- Données démographiques des clients utilisées pour :

- segmentation marketing

- ciblage des campagnes

- analyse du comportement d’achat

### Colonnes clés
- customer_id (identifiant unique)

- country, region

- annual_income

- date_of_birth

### Volumes et périodes
- Volume : ~5000 lignes

- Données historiques (pas de notion temporelle forte sauf DOB)

```
%%sql -r dataframe_1
SELECT COUNT(*) FROM SILVER.CUSTOMER_DEMOGRAPHICS_CLEAN;
```

### Valeurs manquantes / anomalies
- revenus manquants possibles

- incohérences potentielles sur genre ou statut marital

```
%%sql -r dataframe_3
SELECT COUNT(*) FROM SILVER.CUSTOMER_DEMOGRAPHICS_CLEAN WHERE annual_income IS NULL;
```

## 2. CUSTOMER_SERVICE_INTERACTIONS_CLEAN
### Périmètre métier

- Interactions clients avec le service client :

- gestion des réclamations

- satisfaction client

- expérience utilisateur


### Colonnes clés
- interaction_id

- interaction_date

- issue_category

- customer_satisfaction

```
%%sql -r dataframe_9
SELECT MIN(interaction_date), MAX(interaction_date) FROM SILVER.CUSTOMER_SERVICE_INTERACTIONS_CLEAN;
```

### Anomalies

- valeurs nulles dans satisfaction

- durées négatives (si non filtrées)

- incohérences statut (Pending + satisfaction élevée)
